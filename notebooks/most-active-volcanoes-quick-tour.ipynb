# Most Active Volcanoes Since 1900, a Quick Tour

486 volcanoes, 4,454 eruptions. This notebook ranks the most active volcanoes, maps every one of them and looks at eruptions year by year.

```python
import glob
import pandas as pd
import matplotlib.pyplot as plt

def find(name):
    return glob.glob(f"/kaggle/input/**/{name}", recursive=True)[0]

volcanoes = pd.read_csv(find("volcanoes_erupted_since_1900.csv"))
countries = pd.read_csv(find("volcanic_activity_by_country_since_1900.csv"))
years = pd.read_csv(find("eruptions_per_year_since_1900.csv"))
volcanoes.head(10)
```

## 1. The 15 Most Active Volcanoes

```python
top = volcanoes.nsmallest(15, "rank").iloc[::-1]
fig, ax = plt.subplots(figsize=(9, 6))
ax.barh(top.volcano + " (" + top.country + ")", top.eruptions_since_1900, color="#f2a93b")
ax.set_xlabel("Eruptions since 1900")
ax.set_title("Most active volcanoes since 1900")
plt.tight_layout()
plt.show()
```

## 2. Where They Are

Dot size follows the number of eruptions, color the strongest eruption (VEI).

```python
fig, ax = plt.subplots(figsize=(12, 6))
sc = ax.scatter(volcanoes.longitude, volcanoes.latitude, s=volcanoes.eruptions_since_1900 * 3 + 5,
                c=volcanoes.max_vei_since_1900.fillna(0), cmap="YlOrRd", edgecolor="k", linewidth=0.3, alpha=0.85)
ax.set_xlim(-180, 180)
ax.set_ylim(-80, 85)
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
ax.set_title("Volcanoes with an eruption since 1900")
fig.colorbar(sc, label="Strongest eruption since 1900 (VEI)")
plt.show()
```

## 3. Eruptions per Year

2026 is a partial year (to 12 October).

```python
fig, ax = plt.subplots(figsize=(12, 4))
ax.bar(years.year, years.eruptions_started, color="#f2a93b")
ax.set_ylabel("Eruptions started")
ax.set_title("Eruptions started each year, 1900-2026")
plt.show()
years.nlargest(5, "eruptions_started")
```
